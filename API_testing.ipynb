```python
import pandas as pd
import pandas as pd
from openai import OpenAI
import sys
```

```python
sample_data = df_upd.head(5000).to_csv(index = False, sep='|')
```

```python
display(df_upd)
```

```python
content_size = sys.getsizeof(sample_data) / (1024 * 1024)
print(f"Size: {content_size:.2f} MB")
```

```python
#Initial code we used for testing our OpenRouter model
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key="**********************************",
    default_headers={
        "HTTP-Referer": "http://localhost:3000",
        "X-Title": "A&D Fusion",
    }
)

system_instructions = f"""
You are the "A&D Fusion Career Architect." You are working with a curated dataset of 5,000 premium job market entries in Moldova.

--- START OF DATA ---
{sample_data}
--- END OF DATA ---

1. PRECISION: Since we are using a curated subset of 5,000 entries, treat these as the definitive benchmarks for 'Salary Range' and 'Skills'.
2. CV UPGRADE: If a user provides a CV, perform a "Deep Audit":
   - Find the closest 2-3 roles in the provided data.
   - Use the 'skills' and 'Experience' from those roles to suggest 2 high-impact additions to the user's CV.
   - Rewrite a section of their CV using the STAR method (Action -> Result).
3. THE CITY CONTEXT: Use the 'City' data to ensure the advice is localized.
4. LIMITATIONS: If the user's career path isn't represented in these 5,000 rows, honestly state: "Based on our current high-density market subset, we don't have a direct match, but here is the closest strategic advice."
"""

response = client.chat.completions.create(
    model="openrouter/owl-alpha",
    messages=[
        {"role": "system", "content": system_instructions},
        {"role": "user", "content": 'Here is my CV'}
    ]
)

if response and response.choices:
    print(response.choices[0].message.content)
else:
    print("The API returned nothing. Check your token count or connection.")
    print(response)
```
